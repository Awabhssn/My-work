```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/netflix_titles.csv')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df['director'] = df['director'].fillna('No Director')
df['cast'] = df['cast'].fillna('No Cast')
df['country'] = df['country'].fillna('Unknown')
df.dropna(subset=['date_added', 'rating'], inplace=True)
df.isnull().sum()
```

```python
df['date_added'] = pd.to_datetime(df['date_added'].str.strip())
df['year_added'] = df['date_added'].dt.year
df['month_added'] = df['date_added'].dt.month_name()
print(df[['title', 'year_added', 'month_added']].head())
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='type', data=df, palette='Reds')
plt.title('Comparison between Movies and TV Shows')
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.countplot(x='year_added', data=df, order=df['year_added'].value_counts().index.sort_values())
plt.xticks(rotation=45)
plt.title('Content Added Each Year')
plt.show()
```

```python
top_countries = df['country'].value_counts().head(10)

plt.figure(figsize=(10,6))
sns.barplot(x=top_countries.values, y=top_countries.index, palette='viridis')
plt.title('Top 10 Countries with Most Content')
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.countplot(x='rating', data=df, order=df['rating'].value_counts().index, palette='magma')
plt.title('Content Distribution by Rating')
plt.xticks(rotation=45)
plt.show()
```

```python

movies_df = df[df['type'] == 'Movie'].copy()
movies_df['duration_min'] = movies_df['duration'].str.replace(' min', '').astype(float)
plt.figure(figsize=(12,6))
sns.scatterplot(x='release_year', y='duration_min', data=movies_df, alpha=0.5, color='red')
plt.title('Movie Duration Trends over the Years')
plt.show()
```
