```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/DelayedFlights.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
df["ArrTime"] = df["ArrTime"].fillna(df["ArrTime"].mean())
df["DepTime"] = df["DepTime"].fillna(df["DepTime"].mean())
df = df.fillna(df.mean(numeric_only=True))
```

```python
df.isnull().sum()
```

```python
df['ArrDelay'].value_counts()
```

```python
import matplotlib.pyplot as plt

df['ArrDelay'].hist()
plt.title("Distribution of Flight Delays")
plt.xlabel("Delay (minutes)")
plt.ylabel("Count")
plt.show()
```

```python
df.boxplot(column='ArrDelay', by='UniqueCarrier')
```

```python
df.to_csv("CleanedFlights.csv", index=False)
```
